```python
import pandas as pd
from glycowork.motif.annotate import annotate_dataset, annotate_glycan
```

```python
glycan_data = pd.read_csv("../../data/benchmark/glydb_data.csv")
motif_data = pd.read_csv("../../data/benchmark/glygen_motifs.csv")
```

```python
glycans = glycan_data["glycan_structure"]
glycans = [g[:-4] for g in glycans]
glycans[:5]
```

```python
motif_data
```

```python
not_whole_motif_data = motif_data[motif_data["alignment"] != "whole"]
motifs = not_whole_motif_data["iupac"]
motifs = [m[:-4] for m in motifs]
terminals = not_whole_motif_data["alignment"]
terminals = terminals.replace({"substructure": "flexible", "core": "internal"})
```

```python
df = annotate_dataset(
    glycans,
    feature_set=["custom"],
    custom_motifs=motifs,
    termini_list=terminals.tolist()
)

df
```

```python
df.to_csv("../../data/benchmark/glycowork_motif_annotations.csv", index=True)
```
