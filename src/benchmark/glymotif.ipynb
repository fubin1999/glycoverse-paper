```python
import pandas as pd
from glycowork.motif.annotate import annotate_dataset, annotate_glycan
```

```python
glycan_data = pd.read_csv("../../data/benchmark/glycans_glycowork.csv")
motif_data = pd.read_csv("../../data/benchmark/motifs_glycowork.csv")

# glycowork denotes an unknown modification position with O, for example GalOS.
assert not glycan_data["glycan"].str.contains("?S", regex=False).any()
assert not motif_data["motif"].str.contains("?S", regex=False).any()
```

```python
glycans = glycan_data["glycan"]
glycans[:5]
```

```python
motifs = motif_data["motif"]
motifs[:5]
```

```python
df = annotate_dataset(
    glycans,
    feature_set=["custom"],
    custom_motifs=motifs,
)

df
```

```python
df.to_csv(
    "../../data/benchmark/glycowork_motif_annotations.csv",
    index=True,
    index_label="glycowork_glycan",
)
```

```python
sulfate_wildcard_checks = pd.DataFrame(
    {
        "glycan": ["Gal3S", "Gal3/6S", "GalOS"],
        "motif": ["GalOS", "GalOS", "GalOS"],
    }
)
sulfate_wildcard_checks["matches"] = [
    annotate_dataset([glycan], custom_motifs=[motif], feature_set=["custom"]).iloc[0, 0]
    for glycan, motif in sulfate_wildcard_checks[["glycan", "motif"]].itertuples(index=False, name=None)
]
assert sulfate_wildcard_checks["matches"].eq(1).all()
sulfate_wildcard_checks
```
